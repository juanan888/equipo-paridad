```python
import pandas as pd
import altair as alt

# 1. Cargar la base de datos
df = pd.read_excel('BASE DE DATOS.xlsx', sheet_name='Hoja1')

# 2. Procesar los datos (Agrupar por año y sumar pruebas)
df_agrupado = df.groupby('AÑO')[['N°PRUEBAS HOMBRES', 'N° PRUEBAS MUJERES', 'N° PRUEBAS MIXTAS', 'N° PRUEBAS ABIERTAS']].sum().reset_index()

# Transformar los datos a formato largo (melt)
df_largo = df_agrupado.melt(
    id_vars='AÑO',
    value_vars=['N°PRUEBAS HOMBRES', 'N° PRUEBAS MUJERES', 'N° PRUEBAS MIXTAS', 'N° PRUEBAS ABIERTAS'],
    var_name='Tipo de Prueba',
    value_name='Cantidad'
)

# Renombrar para que la leyenda se vea limpia
df_largo['Tipo de Prueba'] = df_largo['Tipo de Prueba'].map({
    'N°PRUEBAS HOMBRES': 'Masculinas',
    'N° PRUEBAS MUJERES': 'Femeninas',
    'N° PRUEBAS MIXTAS': 'Mixtas',
    'N° PRUEBAS ABIERTAS': 'Abiertas'
})

# 3. Visualización: Gráfico de Áreas Apiladas Absoluto
grafico = alt.Chart(df_largo).mark_area(opacity=0.85).encode(
    x=alt.X('AÑO:O', title='Año de los Juegos Olímpicos', axis=alt.Axis(labelAngle=-45)),
    # stack=True mantiene los valores absolutos, apilando las áreas una sobre otra
    y=alt.Y('Cantidad:Q', stack=True, title='Número Absoluto de Pruebas (Medallas)'),
    color=alt.Color('Tipo de Prueba:N',
                    scale=alt.Scale(domain=['Masculinas', 'Femeninas', 'Mixtas', 'Abiertas'],
                                    range=['#1f77b4', '#d62728', '#2ca02c', '#ff7f0e']),
                    title='Categoría de la Prueba'),
    tooltip=['AÑO', 'Tipo de Prueba', 'Cantidad']
).properties(
    title='El techo del programa olímpico: Contención y redistribución de pruebas (1900 - 2024)',
    width=750,
    height=450
)

# 4. Mostrar y Exportar
grafico.display()
grafico.save('vis_01.html')
```
